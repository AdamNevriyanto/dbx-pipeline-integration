## DAB Bundle Commands

Serverless compute does **not** support the Databricks CLI in notebook cells.  
Use one of these alternatives:

### Option 1: Web Terminal (in Databricks UI)
Open the terminal from the sidebar, then run:
```sh
cd /Workspace/Users/adam.nevriyanto@mii.co.id/feature-adam_dbx-pipeline-integration
databricks bundle validate --target personal
databricks bundle deploy --target personal
databricks bundle run --target personal pipeline_btn_development
```

### Option 2: Local Machine
Install the CLI (`pip install databricks-cli`), then run the same commands above.

### Option 3: Ask the Chat Assistant
The chat can run CLI commands for you — just ask:
> "validate the bundle" / "deploy the bundle" / "run the pipeline"
